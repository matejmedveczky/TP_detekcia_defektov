```python
#automatická zmena .xml súborov na .txt

import xml.etree.ElementTree as ET
from pathlib import Path

# mapovanie tried na class_id
CLASS_MAP = {
    "crazing": 0,
    "inclusion": 1,
    "patches": 2,
    "pitted_surface": 3,
    "rolled-in_scale": 4,
    "scratches": 5
}

def convert_box(img_width, img_height, xmin, ymin, xmax, ymax):
    x_center = ((xmin + xmax) / 2) / img_width
    y_center = ((ymin + ymax) / 2) / img_height
    box_width = (xmax - xmin) / img_width
    box_height = (ymax - ymin) / img_height
    return x_center, y_center, box_width, box_height

def convert_xml_file(xml_path):
    tree = ET.parse(xml_path)
    root = tree.getroot()

    size = root.find("size")
    img_width = int(size.find("width").text)
    img_height = int(size.find("height").text)

    yolo_lines = []

    for obj in root.findall("object"):
        class_name = obj.find("name").text.strip()

        if class_name not in CLASS_MAP:
            print(f"Preskakujem neznámu triedu v {xml_path.name}: {class_name}")
            continue

        class_id = CLASS_MAP[class_name]

        bndbox = obj.find("bndbox")
        xmin = float(bndbox.find("xmin").text)
        ymin = float(bndbox.find("ymin").text)
        xmax = float(bndbox.find("xmax").text)
        ymax = float(bndbox.find("ymax").text)

        x_center, y_center, box_width, box_height = convert_box(
            img_width, img_height, xmin, ymin, xmax, ymax
        )

        yolo_lines.append(
            f"{class_id} {x_center:.6f} {y_center:.6f} {box_width:.6f} {box_height:.6f}"
        )

    txt_path = xml_path.with_suffix(".txt")

    with open(txt_path, "w", encoding="utf-8") as f:
        f.write("\n".join(yolo_lines))

    print(f"Hotovo: {xml_path.name} -> {txt_path.name}")
```

```python
from pathlib import Path

label_dirs = [
    Path("datasets/NEU-DET/labels/train"),
    Path("datasets/NEU-DET/labels/val")
]

for label_dir in label_dirs:
    if not label_dir.exists():
        print(f"Priečinok neexistuje: {label_dir}")
        continue

    xml_files = list(label_dir.glob("*.xml"))
    print(f"\nNašiel som {len(xml_files)} XML súborov v: {label_dir}")

    for xml_file in xml_files:
        convert_xml_file(xml_file)

print("\nKonverzia dokončená.")
```
