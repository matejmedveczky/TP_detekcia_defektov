```python
#Výpis či je torch nainštalovaný aj pre grafiku a nie iba pre cpu

import torch
print(torch.__version__)
print(torch.cuda.is_available())
print(torch.cuda.device_count())
print(torch.cuda.get_device_name(0) if torch.cuda.is_available() else "GPU not found")
```

```python
#Trénovanie

from ultralytics import YOLO

model = YOLO("yolo11n.pt")

results = model.train(
    data="TP26_detection_v2-1/data.yaml",
    epochs=200,
    imgsz=512,
    batch=16,
    device=0,
    patience=30,
    project="detectTP",
    name="yolo_tp_dataset_v1_n",
    workers=0,
    pretrained=True
)
```

```python
#Validácia

from ultralytics import YOLO

model = YOLO("runs/detect/detectTP/yolo_tp_dataset_v2/weights/best.pt")

metrics = model.val(
    data="TP26_detection_v2-1/data.yaml",
    project="validationTP",
    name="val_TP_dataset_v1_n",
)

print("mAP50-95:", metrics.box.map)
print("mAP50:", metrics.box.map50)
print("Precision:", metrics.box.mp)
print("Recall:", metrics.box.mr)
```

```python
#testovanie

from ultralytics import YOLO

model = YOLO("runs/detect/detectTP/yolo_tp_dataset_v2/weights/best.pt")

results = model.predict(
    source="TP26_detection_v2-1/test/images",
    save=True,
    device=0,
    conf=0.5,
    iou=0.5,
    project="predictTP",
    name="predict_TP_dataset_v2",
    exist_ok=True
)
```

```python
#Vykreslovanie metrík

from IPython.display import HTML, display

def show_centered_image(path, width=800):
    display(HTML(f"""
    <div style="text-align: center;">
        <img src="{path}" width="{width}">
    </div>
    """))

show_centered_image(
    "runs/detect/detectTP/yolo_tp_dataset_v2/confusion_matrix.png",
    width=1000,
)

show_centered_image(
    "runs/detect/detectTP/yolo_tp_dataset_v2/confusion_matrix_normalized.png",
    width=1000,
)

show_centered_image(
    "runs/detect/detectTP/yolo_tp_dataset_v2/results.png",
    width=1600,
)
```
