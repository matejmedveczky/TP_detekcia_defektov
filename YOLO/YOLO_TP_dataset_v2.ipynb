```python
from ultralytics import YOLO

model = YOLO("yolo11x.pt")

results = model.train(
    data="TP26_detection_v2-1/data.yaml",
    epochs=200,
    imgsz=512,
    batch=8,
    project="detectTP",
    name="yolo_tp_dataset_v2_x",
    device=0,
    workers=0,
    patience=50,
    cache=True,
    pretrained=True,
    optimizer="auto",
    cos_lr=True,
    close_mosaic=10,
    amp=True
)
```

```python
#test na obrázkoch tp26_detection-1/test

from ultralytics import YOLO

model = YOLO("runs/detect/detectTP/yolo_tp_dataset_v2_x/weights/best.pt")

results = model.predict(
    source="TP26_detection_v2-1/test/images",
    save=True,
    device=0,
    project="predictTP",
    name="predict_TP_dataset_v2_x",
    exist_ok=True
)
```
