```python
from ultralytics import YOLO
from pathlib import Path

# načítanie natrenovaného modelu
model = YOLO("runs/classify/runs/classify_clean/neu_test/weights/best.pt")

# cesta na validačný dataset
val_dir = Path("datasets/NEU-DET/images/validation")

correct = 0
total = 0

# prejdi všetky triedy
for class_dir in val_dir.iterdir():
    if not class_dir.is_dir():
        continue

    true_class = class_dir.name

    # prejdi všetky obrázky v danej triede
    for img_path in class_dir.iterdir():
        if not img_path.is_file():
            continue

        results = model.predict(str(img_path), verbose=False)
        pred_class = results[0].names[results[0].probs.top1]

        if pred_class == true_class:
            correct += 1

        total += 1

accuracy = correct / total * 100
print(f"Správne: {correct}/{total}")
print(f"Accuracy: {accuracy:.2f}%")
```
