```python
from ultralytics import YOLO

model = YOLO("yolo11n.pt")

results = model.train(
    data="TP26_detection_v2-3/data.yaml",  # Updated to use new Roboflow dataset
    batch=16,
    optimizer='AdamW',
    imgsz=512,
    lr0=0.001,
    lrf=0.01,
    patience=30,
    epochs=100,
    weight_decay=0.001,   # up from 0.0005, more regularization
    close_mosaic=20,      # stop mosaic augmentation earlier    
    project="detectTP",
    name="yolo_metal-2",
    exist_ok=True,
)
```

```python

# Load and verify the Roboflow dataset
import os
from pathlib import Path

dataset_path = Path("TP26_detection_v2-3")
data_yaml = dataset_path / "data.yaml"

print(f"Dataset path: {dataset_path.absolute()}")
print(f"data.yaml exists: {data_yaml.exists()}")
print(f"\nDataset structure:")
for item in dataset_path.iterdir():
    print(f"  {item.name}")
```
